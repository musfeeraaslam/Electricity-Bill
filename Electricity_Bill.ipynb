```python
bill= float(input("Enter the monthly electricity bill:"))
threshold= 20000
surcharge_rate= 0.10
if bill > threshold:
  surcharge= bill*surcharge_rate  #This finds how much extra needs to be charged.
  final_bill= bill+ surcharge_rate  #This gives the total amount the household must pay.
  print(f"Bill exceed Rs.{threshold} --- surcharge applied")
  print(f"Original bill is: Rs.{bill:.2f}")
  print(f"Surcharge (10%): {surcharge:.2f}")
  print(f"Final Bill      : Rs. {final_bill:.2f}")
else:
  print(f"Bill does not exceed: Rs. {threshold} --- no surcharge apply")
  print(f"Final bill is: Rs. {bill:.2f}")
```
